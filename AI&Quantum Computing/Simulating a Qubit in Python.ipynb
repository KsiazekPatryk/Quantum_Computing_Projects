```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create a Quantum Circuit with 1 qubit and 1 classical bit 
qc = QuantumCircuit(1, 1)

# Apply Hadamard gate to create superposition
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Use Quiskit's Aer Simulator
simulator = Aer.get_backend('aer_simulator')

# Run the circuit on the simulator
job = simulator.run(qc, shots=1000)     # 1000 measurements
result = job.result()

# Get the measurement results
counts = result.get_counts(qc)
print(counts)

# Convert dictionary keys and values into lists
labels = list(counts.keys())    #['0', '1']
values = list(counts.values())  # [490, 510]

# Define colors for the bars
colors = ['blue', 'red']

# Plot the bar chart
plt.bar(labels, values, color=colors)

#Add labels and title
plt.xlabel('Measurement Outcome')
plt.ylabel('Counts')
plt.title('Measurement Results of Quantum Circuit')

# Display the plot
plt.show()


```
