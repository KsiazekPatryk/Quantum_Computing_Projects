```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer

#Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1,1)

# Apply HAdamard gate to create superposition
qc.h(0)

# Measure the qubit
qc.measure(0,0)

# Draw the circuit
qc.draw('mpl')

# Use Qiskit's Aer Simulator
simulator = Aer.get_backend('aer_simulator')

# Run the circuit on the simulator
job = simulator.run(qc, shots=1000)
result = job.result()

# Get the measurement results
counts = result.get_counts(qc)
print(counts)
```

```python
# Implement a Bell State Circuit (Entanglement)

qc = QuantumCircuit(2) # Two qubits circuit


# Apply HAdamard gate to create superposition
qc.h(0)

# Apply a CNOT (Controlled-NOT) gate to entangle the qubits

qc.cx(0, 1)

# Measure all qubits
qc.measure_all()


# Draw the circuit
qc.draw('mpl')


# Use Qiskit's Aer Simulator
simulator = Aer.get_backend('aer_simulator')

# Run the circuit on the simulator
job = simulator.run(qc, shots=1000)
result = job.result()

# Get the measurement results
counts = result.get_counts(qc)
print(counts)
```
