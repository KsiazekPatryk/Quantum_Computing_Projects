LEGB

L -> Local
E -> Enclosing
G -> Global
B -> Built-in

```python
my_string = "Atil"
#Global

def my_func():
    my_string = "James"
    #Enclosing

    def my_func_2():

        #Local
        my_string = "Lars"
        print(my_string)
    
    my_func_2()
```

```python
my_func()
```

```python
my_string
```
