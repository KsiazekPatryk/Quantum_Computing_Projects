```python
def func(new_func):
    print("func started")
    new_func()
    print("func ended")
```

```python
def hello_func():
    print("hello world")
```

```python
func(hello_func)
```

```python
def new_func():
    print("new func")
    def new_func_2():
        print("new func 2")
    return new_func_2
```

```python
new_func()
```

```python
new_string = new_func()
```

```python
new_string()
```

```python
def decorator_function(func):
    def wrapper_function():
        print("wrappper started")
        func()
        print("wrapper stopped")
    return wrapper_function
```

```python
def func_new():
    print("Hello World")
```

```python
example_function = decorator_function(func_new)
```

```python
example_function()
```

```python
@decorator_function
def func_new():
    print("Hello World")
```

```python
func_new()
```
