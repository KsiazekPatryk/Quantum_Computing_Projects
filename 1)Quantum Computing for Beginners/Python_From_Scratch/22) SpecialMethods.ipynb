```python
class Fruits():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f"{self.name} has {self.calories}"
    
    def __len__(self):
        return self.calories
```

```python
my_fruit = Fruits("Banana", 200)
```

```python
print(my_fruit)
```
