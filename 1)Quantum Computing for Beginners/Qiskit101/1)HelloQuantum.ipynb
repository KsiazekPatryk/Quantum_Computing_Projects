```python
from qiskit import *
```

```python
circuit = QuantumCircuit(2,2)
```

```python
#quantum_register = QuantumRegister(2)
#classical_register = ClassicalRegister(2)
#circuit = QuantumCircuit(quantum_register, classical_register)
```

```python
circuit.draw()
```

```python
%matplotlib inline
```

```python
circuit.draw(output='mpl')
```

```python
circuit.h(0)
circuit.draw(output = 'mpl')
```

```python
circuit.cx(0,1) # 0-> contol,qubit, 1-> target qubit
circuit.measure([0,1] , [0,1])
circuit.draw(outpu='mpl')
```

```python
simulator = Aer.get_backend('qasm_simulator')
```

```python
result = execute(circuit,backend = simulator).result()
```

```python
from qiskit.visualization import plot_histogram
```

```python
plot_histogram(result. get_counst(circuit))
```

```python
IBMQ.load_account
```

```python
provider = IBMQ.get_provider("ibm-q")
```

```python
quantum_computer = provider.get_backend('ibmq_lima')
```

```python
import qiskit.tools.jupyter
%qiskit_job_watcher

job = execute(circuit,backend = quantum_computer)
```

```python
from qiskit.tools.monitor import job_monitor
job_monitor(job)
```

```python
qunatum_result = job.result()
plot_histogram(qunatum_result.get_counts(circuit))
```
