```python
from qiskit import *
```

```python
%matplotlib inline
from qiskit.tools.visualization import plot_histogram
```

```python
secretNumber = '10101010100'
```

```python
circuit = QuantumCircuit(len(secretNumber), len(secretNumber))
```

```python
circuit.h(range(len(secretNumber)))
circuit.x(len(secretNumber))

circuit.barrier()

for index, one in enumerate(reversed(secretNumber)):
    print(f"index"{index} is {one}")
    if one == "1":
        circuit.cx(index,len(secretNumber))

circuit.barrier()

circuit.h(range(len(secretNumber)))

circuit.barrier()

circuit.measure(range(len(secretNumber)), range(len(secretNumber)))

```

```python
circuit.drwa(output='mpl')
```

```python
simulator = Aer.get_backend
result = execute(circuit, backend= simulator, shots=1).result()
counts = result.get_counts()
print(counts)
```

```python
plot_histogram(counts)
```
